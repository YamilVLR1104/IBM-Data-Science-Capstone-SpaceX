# SpaceX Falcon 9 First Stage Landing Prediction
## Lab 1: Collecting the Data via SpaceX REST API

```python
import requests
import pandas as pd
import numpy as np
import datetime

spacex_url = "https://api.spacexdata.com/v4/launches/past"
response = requests.get(spacex_url)
data = pd.json_normalize(response.json())
print("First row static_fire_date_utc:", data.loc[0, 'static_fire_date_utc'])
print("Year in first row:", str(data.loc[0, 'static_fire_date_utc'])[:4])
```

```python
# Filter single core, single payload, date <= 2020-11-13, and remove Falcon 1
data = data[data['cores'].map(len) == 1]
data = data[data['payloads'].map(len) == 1]
data['cores'] = data['cores'].map(lambda x: x[0])
data['payloads'] = data['payloads'].map(lambda x: x[0])
data['date'] = pd.to_datetime(data['date_utc']).dt.date
data = data[data['date'] <= datetime.date(2020, 11, 13)]

# After enriching with rocket, launchpad, payload, and core endpoints:
data_falcon9 = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv")
print("Falcon 9 launches count:", len(data_falcon9))
print("Missing values per column:\n", data_falcon9.isnull().sum())
data_falcon9.to_csv('dataset_part_1.csv', index=False)
```
