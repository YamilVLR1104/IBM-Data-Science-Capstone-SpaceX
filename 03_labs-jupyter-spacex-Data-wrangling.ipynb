# SpaceX Falcon 9 First Stage Landing Prediction
## Lab 3: Data Wrangling & Exploratory Analysis

```python
import pandas as pd
import numpy as np

df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv")
print("Launches per LaunchSite:\n", df['LaunchSite'].value_counts())
print("\nLaunches per Orbit:\n", df['Orbit'].value_counts())
landing_outcomes = df['Outcome'].value_counts()
print("\nLanding Outcomes:\n", landing_outcomes)
```

```python
bad_outcomes = set(landing_outcomes.keys()[[1, 3, 5, 6, 7]])
df['Class'] = [0 if outcome in bad_outcomes else 1 for outcome in df['Outcome']]
print(f"Overall landing success rate: {df['Class'].mean():.4f} ({df['Class'].mean()*100:.2f}%)")
df.to_csv("dataset_part_2.csv", index=False)
```
